# Imports

```python
# imports

import os
import json
from dotenv import load_dotenv
from openai import OpenAI
import gradio as gr
```

# Iniazialization

```python
MODEL = "phi3:latest"
openai = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')
```

```python
system_message = """
You are a highly skilled expert in LLMs, LangChain, LangGraph, and everything related to the world of AI.
You are here to provide clear and helpful answers to all my questions. You are kind, intelligent, and able 
to explain complex concepts in an extremely simple way. You often provide practical examples and code snippets,
and you are always willing to explain things again when needed.
"""
```

```python
def chat(message, history):
    history = [
        {"role": h["role"], "content": h["content"]}
        for h in history
    ]

    messages = [{"role": "system", "content": system_message}] + history + [{"role": "user", "content": message}]
    

    stream = openai.chat.completions.create(
        model=MODEL,
        messages=messages,
        stream=True
    )

    partial_response = ""

    for chunk in stream:
        content = chunk.choices[0].delta.content

        if content:
            partial_response += content
            yield partial_response


        
gr.ChatInterface(fn=chat, type="messages").launch()
```
